```python
katet_1 = int(input('Введите длину катета:'))
katet_2 = int(input('Введите длину катета:'))
gip = int(input('Введите длину гипотенузы:'))
a = mt.asin(katet_1 / gip)
b = mt.asin(katet_2 / gip)
c1 = round(min(a, b), 3)
print('Минимум=', c1)
c2 = round(max(a, b), 3)
print('Максимум=', c2)
d1 = c1 * 180 / 3.14
d2 = c2 * 180 / 3.14
e1 = round(min(d1, d2), 3)
e2 = round(max(d1, d2), 3)
print('Наименьшее значение угла=', e1)
print('Наибольшее значение угла=', e2)
```
